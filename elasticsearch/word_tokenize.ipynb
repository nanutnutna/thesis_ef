```python
import pandas as pd
import re

df = pd.read_json("combine_20250629.json")
sentence = df['Name']

def has_thai(text):
    thai_pattern = r'[ก-๙]'
    return bool(re.search(thai_pattern, str(text)))

for sent in sentence:
    if has_thai(sent):
        print(sent.split()[0])
    else:
        print(sent)
```

```python
from pythainlp import word_tokenize

text = "ก๊าซปิโตรเลียมเหลวมีราคาแพง"

# ลอง engine ต่างๆ
engines = ['newmm', 'longest', 'attacut']

for engine in engines:
    try:
        tokens = word_tokenize(text, engine=engine)
        print(f"{engine}: {tokens}")
    except:
        print(f"{engine}: ไม่สามารถใช้งานได้")
```
