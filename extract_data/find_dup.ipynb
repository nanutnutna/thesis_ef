```python
import pandas as pd

def clean_duplicate_words_anywhere(df, target_word):
   """
   ลบคำซ้ำจากทั้ง DataFrame 
   - เก็บแถวแรกที่มีคำที่ต้องการ และแก้ไขให้เก็บคำนั้นตัวแรกเท่านั้น
   - ลบแถวอื่นๆ ที่มีคำนั้นออกทั้งหมด
   - เก็บแถวที่ไม่มีคำนั้นไว้
   """
   
   # ฟังก์ชันตรวจสอบว่าแถวมีคำที่ต้องการหรือไม่
   def has_word_in_row(row):
       for value in row:
           if pd.notna(value) and target_word in str(value).split():
               return True
       return False
   
   # ฟังก์ชันลบคำซ้ำในแถว เก็บตัวแรกไว้
   def clean_row(row):
       first_found = False
       cleaned_row = row.copy()
       
       for col_name in row.index:
           if pd.notna(row[col_name]):
               text = str(row[col_name])
               words = text.split()
               result = []
               
               for word in words:
                   if word == target_word:
                       if not first_found:
                           result.append(word)
                           first_found = True
                       # ข้ามคำที่ซ้ำ
                   else:
                       result.append(word)
               
               cleaned_row[col_name] = ' '.join(result)
       
       return cleaned_row
   
   # หาแถวที่มีคำที่ต้องการ
   has_word_mask = df.apply(has_word_in_row, axis=1)
   
   if has_word_mask.any():
       # หาแถวแรกที่มีคำที่ต้องการ
       first_word_index = df[has_word_mask].index[0]
       
       # สร้าง copy ของ DataFrame
       df_result = df.copy()
       
       # แก้ไขแถวแรก - เก็บคำที่ต้องการตัวแรกเท่านั้น
       df_result.loc[first_word_index] = clean_row(df_result.loc[first_word_index])
       
       # เก็บเฉพาะแถวที่ต้องการ (แถวแรกที่มีคำ + แถวที่ไม่มีคำนั้น)
       mask = (df_result.index == first_word_index) | (~has_word_mask)
       return df_result[mask]
   else:
       return df  # ถ้าไม่มีคำที่ต้องการเลย

# ตัวอย่างการใช้งาน
sample_data = {
   'col1': ['กรดน้ำส้ม ดาน อินทรีย์', 'สารเคมี', 'เอทานอล ตัวทำละลาย', 'กรด ดาน', 'ดาน เคมี'],
   'col2': ['เคมี อนินทรีย์', 'ดาน ตัวทำละลาย ดาน', 'อินทรีย์', 'สาร ดาน อนินทรีย์', 'น้ำ'],
   'col3': ['ตัวทำละลาย', 'อนินทรีย์ ดาน', 'เคมี ดาน ดาน', 'สาร', 'ตัวทำละลาย ดาน ดาน ดาน']
}

df = pd.DataFrame(sample_data)
print("ข้อมูลเดิม:")
print(df)

# ใช้งาน
df_cleaned = clean_duplicate_words_anywhere(df, 'ดาน')
print("\nหลังทำความสะอาดคำ 'ดาน':")
print(df_cleaned)

# ใช้กับคำอื่น
df_cleaned2 = clean_duplicate_words_anywhere(df, 'เคมี')
print("\nหลังทำความสะอาดคำ 'เคมี':")
print(df_cleaned2)

df_cleaned2
```

```python
import pandas as pd

def clean_multiple_words_anywhere(df, word_list):
   """
   ลบคำซ้ำหลายคำพร้อมกันจากทั้ง DataFrame 
   - เก็บแถวแรกที่มีแต่ละคำ และแก้ไขให้เก็บคำนั้นตัวแรกเท่านั้น
   - ลบแถวอื่นๆ ที่มีคำเหล่านั้นออกทั้งหมด
   - เก็บแถวที่ไม่มีคำเหล่านั้นไว้
   """
   
   # ฟังก์ชันตรวจสอบว่าแถวมีคำใดๆ ในรายการหรือไม่
   def has_any_word_in_row(row, words):
       for value in row:
           if pd.notna(value):
               text_words = str(value).split()
               for word in words:
                   if word in text_words:
                       return True, word
       return False, None
   
   # ฟังก์ชันลบคำซ้ำทั้งหมดในแถว เก็บตัวแรกของแต่ละคำไว้
   def clean_row_multiple_words(row, words):
       word_found = {word: False for word in words}
       cleaned_row = row.copy()
       
       for col_name in row.index:
           if pd.notna(row[col_name]):
               text = str(row[col_name])
               text_words = text.split()
               result = []
               
               for word in text_words:
                   if word in words:
                       if not word_found[word]:
                           result.append(word)
                           word_found[word] = True
                       # ข้ามคำที่ซ้ำ
                   else:
                       result.append(word)
               
               cleaned_row[col_name] = ' '.join(result)
       
       return cleaned_row
   
   # หาแถวที่มีคำใดๆ ในรายการ
   rows_with_words = {}
   for idx, row in df.iterrows():
       has_word, found_word = has_any_word_in_row(row, word_list)
       if has_word:
           if found_word not in rows_with_words:
               rows_with_words[found_word] = idx
   
   if rows_with_words:
       # สร้าง copy ของ DataFrame
       df_result = df.copy()
       
       # แก้ไขแถวแรกของแต่ละคำ
       for word, first_idx in rows_with_words.items():
           df_result.loc[first_idx] = clean_row_multiple_words(df_result.loc[first_idx], [word])
       
       # หาแถวที่ต้องเก็บ (แถวแรกของแต่ละคำ + แถวที่ไม่มีคำใดๆ)
       keep_indices = set(rows_with_words.values())
       
       for idx, row in df.iterrows():
           has_word, _ = has_any_word_in_row(row, word_list)
           if not has_word:
               keep_indices.add(idx)
       
       return df_result.loc[sorted(keep_indices)]
   else:
       return df  # ถ้าไม่มีคำใดๆ ในรายการเลย

# ตัวอย่างการใช้งาน
sample_data = {
   'col1': ['กรดน้ำส้ม ดาน อินทรีย์', 'สารเคมี', 'เอทานอล ตัวทำละลาย', 'กรด ดาน', 'ดาน เคมี', 'อินทรีย์ เคมี'],
   'col2': ['เคมี อนินทรีย์', 'ดาน ตัวทำละลาย ดาน', 'อินทรีย์', 'สาร ดาน อนินทรีย์', 'น้ำ', 'เคมี เคมี'],
   'col3': ['ตัวทำละลาย', 'อนินทรีย์ ดาน', 'เคมี ดาน ดาน', 'สาร', 'ตัวทำละลาย ดาน', 'อินทรีย์ อินทรีย์']
}

df = pd.DataFrame(sample_data)
print("ข้อมูลเดิม:")
print(df)

# ใช้งานกับหลายคำพร้อมกัน
words_to_clean = ['ดาน', 'เคมี', 'อินทรีย์']
df_cleaned = clean_multiple_words_anywhere(df, words_to_clean)
print(f"\nหลังทำความสะอาดคำ {words_to_clean}:")
print(df_cleaned)
```
