```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score as sil_score
import numpy as np

df = pd.read_csv('inner_join_20250526.csv')
industries = df['Industrials'].unique()
names = df['Name'].unique()
# print(industries)

vectorizer = TfidfVectorizer(max_features=1000)
X = vectorizer.fit_transform(df['Name'].unique())
inertias = []
silhouette_scores = []
k_range = range(6,10)

for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X)
    inertias.append(kmeans.inertia_)
    silhouette_scores.append(sil_score(X.toarray(), kmeans.labels_))

best_k = k_range[np.argmax(silhouette_scores)]

kmeans = KMeans(n_clusters=best_k, random_state=42)
clusters = kmeans.fit_predict(X)

for i, doc in enumerate(names):
    print(f"name: '{doc}' -> กลุ่ม {clusters[i]}")
```
