```python
def read_synonyms_file(file_path):
    with open(file_path, 'r', encoding='utf-8') as file:
        return [line.strip() for line in file.readlines() if line.strip()]
```

```python
synonyms_path = r"C:\Users\Nattapot\Downloads\synonym_2.txt"
initial_synonyms = read_synonyms_file(synonyms_path)
```

```python
for word in initial_synonyms:
    print(word)
```
