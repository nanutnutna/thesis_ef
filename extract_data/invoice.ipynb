```python
import requests
from bs4 import BeautifulSoup
import urllib3
import pandas as pd
from tqdm import tqdm
import math
from datetime import datetime
import pandasql as ps

currentDate = datetime.now()
file_date = datetime.strftime(currentDate,"%Y%m%d")
```

```python
to_pdf = []
for num in range(1,17):
    url = f'https://wt.go.th/news//?page={num}&cid=12&limit=36'
    response = requests.get(url, verify=False)
    soup = BeautifulSoup(response.content, 'html.parser')
    table = soup.find_all(class_='mid')
    # print(table)
    for i, element in enumerate(table, start=1):
        links = element.find_all('a')  # Find all <h4> inside the current element
        for link in links:
            to_pdf.append(link.get('href'))
```

```python
link_file = []
for i in to_pdf:
    response = requests.get(i, verify=False)
    soup = BeautifulSoup(response.content, 'html.parser')
    table = soup.find_all(class_='name')
    # print(table)
    for i, element in enumerate(table, start=1):
        links = element.find_all('a')  # Find all <h4> inside the current element
        for link in links:
            link_file.append(link.get('href'))
```

```python
import urllib.request
import os

def download_pdf_urllib(url, output_path):
    try:
        # Create directory if it doesn't exist
        os.makedirs(os.path.dirname(os.path.abspath(output_path)), exist_ok=True)
        
        # Download the file
        urllib.request.urlretrieve(url, output_path)
        
        print(f"PDF successfully downloaded to {output_path}")
        return True
        
    except Exception as e:
        print(f"Error downloading PDF: {e}")
        return False
```
